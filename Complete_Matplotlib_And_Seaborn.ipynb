```python
#Bar Chart
x = ['A','B','C','D','E']
y = [10, 20, 55, 35, 45]

plt.bar(x,y)
plt.title('Bar Chart Example')
```

```python
# Histogram
# used for distribution analysis
# data
import random
data = [random.randint(1, 20) for _ in range(500)]

plt.hist(data, bins=20) # histogram chart
plt.title("Histogram Example")
```

```python
#Pie chart
# used to show part-to-whole relationships/data

# data
categories = ['A','B','C','D','E']
sales = [10, 20, 55, 35, 45]

plt.pie(sales, labels=categories, autopct = '%1.1f%%')
plt.title("Pie Chart Example")
```

```python

plt.pie(sales, labels=categories, autopct = '%1.1f%%', startangle=90)   #used to define the starting point
plt.title("Pie Chart Example")
```

```python
# Scatter plot
# used to find relationship btwn variables

# data
y1 = [10, 20, 25, 35, 45]
y2 = [20, 30, 35, 45, 55]
plt.scatter(y1, y2)
plt.title("Scatter Plot Example")
```

```python
#Subplots --> used to show multiple charts in one figure

# data-1 - bar chart
categories = ['Mon','Tue','Wed','Thu','Fri']
sales = [10, 20, 55, 35, 45]

# data-2 - scatter plot
y1 = [10, 20, 25, 35, 45]
y2 = [20, 30, 35, 45, 55]

plt.figure(figsize=(10,4))

# first plot- bar chart
plt.subplot(1,2,1) # row, column, position
plt.bar(categories, sales)
plt.title("Daily Sales")
plt.xlabel("Week Days")
plt.ylabel("Sales")


# second plot- scatter chart
plt.subplot(1,2,2) # row, column, position
plt.scatter(y1, y2)
plt.title("User Example")
plt.xlabel("User1")
plt.ylabel("User2")
```

```python
#Matplotlib with Pandas - real data
# create dataframe
import pandas as pd

# data
data = {
    'Month' : ['Jan', 'Feb', 'Mar', 'Apr'],
    'Sales' : [12000, 11000, 13000, 25000]
}

df = pd.DataFrame(data)
df
```

```python
plt.bar(df['Month'], df['Sales'])
plt.title("Matplotlib with Pandas")
plt.xlabel("Month")
plt.ylabel("Sales")
```

```python
plt.pie(df['Sales'], labels=df['Month'], autopct = '%1.1f%%')
plt.title("Matplotlib with Pandas")
```

```python
plt.plot(df['Month'], df['Sales'],color='green', marker='o', linestyle='--', linewidth=2, markersize=12)
plt.title("Matplotlib with Pandas")
plt.xlabel("Month")
plt.ylabel("Sales")
```

```python
# Save charts
plt.bar(df['Month'], df['Sales'])
plt.title("Matplotlib with Pandas")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.savefig("Monthly_Sales_from_df.png")
```

```python
#Categorical Data Plots
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset("tips")
df
```

```python
sns.countplot( df['sex'])
```

```python
sns.countplot(x = df['sex'])
```

```python
sns.countplot(x = df['sex'],hue = df['smoker'])
```

```python
#barplot
sns.barplot(x = df['sex'],y = df['total_bill'])          #x-axis pr categorical data hoga or y-axis pr quantitative data hoga
```

```python
import numpy as np
sns.barplot(x = df['sex'],y = df['tip'],estimator=np.sum)
```

```python
#Box Plot
sns.boxplot(x= 'tip',y = 'day',data = df,palette = 'rainbow')
```

```python
#violinplot
sns.violinplot(x = 'day',y = 'total_bill',data = df)
```

```python
#stripplot
sns.stripplot(x = 'day',y = 'total_bill',data = df)
```

```python
#swarmplot
sns.swarmplot(x = 'day',y = 'total_bill',data = df)
```

```python
#Combining violinplot and swarmplot
sns.violinplot(x= 'tip',y = 'day',data = df)
sns.swarmplot(x= 'tip',y = 'day',data = df)
```
